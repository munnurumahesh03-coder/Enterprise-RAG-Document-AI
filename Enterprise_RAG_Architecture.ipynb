```python
# Cell 3: Load and Chunk the Document (FIXED)
!pip install -q langchain-text-splitters

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("⏳ Reading the PDF and chunking the text...")

# 1. Load the PDF
loader = PyPDFLoader("attention_paper.pdf")
pages = loader.load()
print(f"📄 Successfully loaded {len(pages)} pages.")

# 2. Initialize the Text Splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

# 3. Chop the pages into chunks
chunks = text_splitter.split_documents(pages)

print(f"✂️ Document successfully split into {len(chunks)} chunks!")
print(f"🔍 Preview of Chunk 1:\n{chunks[0].page_content[:200]}...")
```

```python
# Cell 4: Embeddings and Vector Database (ChromaDB)
!pip install -q langchain-huggingface

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

print("🧠 Downloading Embedding Model (Turning text into math)...")
# We use a fast, open-source embedding model from HuggingFace
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

print("🗄️ Building the Chroma Vector Database...")
# This takes our 52 chunks, embeds them, and stores them   in a local ChromaDB
vector_db = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="./chroma_db"
)

print("✅ Vector Database built successfully! The math is ready to be searched.")
```

```python
# Cell 5: The RAG Pipeline with Langfuse Tracing
import os
import getpass
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langfuse.langchain import CallbackHandler

# 1. Secure API Keys
if "GROQ_API_KEY" not in os.environ:
    print("🔑 Enter your Groq API Key:")
    os.environ["GROQ_API_KEY"] = getpass.getpass()

# (In production, put these in Streamlit Secrets. For Colab, paste your keys here)
os.environ["LANGFUSE_PUBLIC_KEY"] = "your_public_key_here"
os.environ["LANGFUSE_SECRET_KEY"] = "your_secret_key_here"
os.environ["LANGFUSE_HOST"] = "https://us.langfuse.com"

langfuse_handler = CallbackHandler( ) # <-- INITIALIZE TRACING

print("🧠 Waking up the LLM...")
llm = ChatGroq(temperature=0, model_name="qwen/qwen3.8-27b")

# 2. Turn ChromaDB into a Retriever (Fetch top 3 chunks)
retriever = vector_db.as_retriever(search_kwargs={"k": 3})

# 3. The Strict RAG Prompt
template = """
You are an elite AI Research Assistant. Answer the question based ONLY on the following context.
If the answer is not in the context, say "I cannot answer this based on the provided document."
Do not hallucinate.

Context:
{context}

Question: {question}

Answer:
"""
prompt = ChatPromptTemplate.from_template(template)

# 4. Build the LCEL Chain
rag_chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
print("✅ Enterprise RAG Pipeline is LIVE!")

# 5. Ask a highly technical question about the paper!
question = "Why is self-attention better than recurrent neural networks (RNNs) according to the paper?"
print(f"\n🗣️ USER: {question}\n")

# <-- THIS IS WHERE THE MAGIC HAPPENS
answer = rag_chain.invoke(
    question,
    config={"callbacks": [langfuse_handler]}
)

print(f"🎯 FINAL ANSWER:\n{answer}")
```

```python
%%writefile app.py
import streamlit as st
import os
import tempfile
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langfuse.langchain import CallbackHandler # <-- 1. LANGFUSE IMPORT

# --- UI Configuration ---
st.set_page_config(page_title="Enterprise Document AI", page_icon="📄", layout="wide")
st.title("📄 Enterprise Document AI & RAG Engine")
st.markdown("Upload any PDF document and ask questions. The AI will extract the exact context and provide hallucination-free answers.")

# Fetch API keys securely from Streamlit Secrets
GROQ_API_KEY = st.secrets["GROQ_API_KEY"]

# Securely set Langfuse env variables from Streamlit Secrets
os.environ["LANGFUSE_PUBLIC_KEY"] = st.secrets.get("LANGFUSE_PUBLIC_KEY", "dummy_key")
os.environ["LANGFUSE_SECRET_KEY"] = st.secrets.get("LANGFUSE_SECRET_KEY", "dummy_key")
os.environ["LANGFUSE_HOST"] = st.secrets.get("LANGFUSE_HOST", "https://us.langfuse.com" )

# --- Sidebar: Setup & Upload ---
with st.sidebar:
    st.header("📂 Document Upload")
    uploaded_file = st.file_uploader("Upload a PDF", type=["pdf"])

    if st.button("Process Document"):
        if not uploaded_file:
            st.error("⚠️ Please upload a PDF document.")
        else:
            with st.spinner("Chunking and Embedding Document..."):
                # Save the uploaded file to a temporary location
                with tempfile.NamedTemporaryFile(delete=False, suffix=".pdf") as tmp_file:
                    tmp_file.write(uploaded_file.getvalue())
                    tmp_path = tmp_file.name

                # Extract and Chunk
                loader = PyPDFLoader(tmp_path)
                pages = loader.load()
                text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
                chunks = text_splitter.split_documents(pages)

                # Embed and Store in ChromaDB
                embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
                vector_db = Chroma.from_documents(documents=chunks, embedding=embeddings)

                st.session_state.vector_db = vector_db
                st.success("✅ Document processed successfully! You can now chat.")

# --- Helper Function ---
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# --- Main Chat Interface ---
if "messages" not in st.session_state:
    st.session_state.messages = []

# Display chat history
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# User Input
if prompt := st.chat_input("Ask a question about your document..."):
    # Add user message to UI
    st.session_state.messages.append({"role": "user", "content": prompt})
    with st.chat_message("user"):
        st.markdown(prompt)

    # Generate Assistant Response
    with st.chat_message("assistant"):
        if "vector_db" not in st.session_state:
            st.warning("⚠️ Please upload and process a PDF document first.")
        else:
            with st.spinner("Searching document..."):
                # Initialize LLM with deterministic temperature
                llm = ChatGroq(temperature=0, model_name="qwen/qwen3.8-27b", api_key=GROQ_API_KEY, max_tokens=500)

                # Fetch Top-3 similar chunks
                retriever = st.session_state.vector_db.as_retriever(search_kwargs={"k": 3})

                # 2. Initialize the Langfuse Tracer
                langfuse_handler = CallbackHandler()

                # Strict Guardrail Prompt
                template = """
                You are an elite AI Research Assistant. Answer the question based ONLY on the following context.
                If the answer is not in the context, say "I cannot answer this based on the provided document."
                Do not hallucinate.

                Context:
                {context}

                Question: {question}

                Answer:
                """
                prompt_template = ChatPromptTemplate.from_template(template)

                # Build LCEL Chain
                rag_chain = (
                    {"context": retriever | format_docs, "question": RunnablePassthrough()}
                    | prompt_template
                    | llm
                    | StrOutputParser()
                )

                # 3. Inject the Langfuse handler into the execution
                response = rag_chain.invoke(
                    prompt,
                    config={"callbacks": [langfuse_handler]}
                )

                # 4. FORCE SEND TO LANGFUSE DASHBOARD (Crucial for Streamlit)
                langfuse_handler.flush()

                # Display and save response
                st.markdown(response)
                st.session_state.messages.append({"role": "assistant", "content": response})
```

```python
import os
import json
from groq import Groq

# 1. Initialize the Judge LLM
# It securely pulls the key directly from your OS environment variables
judge_client = Groq(api_key=os.environ.get("GROQ_API_KEY"))

def evaluate_rag_hallucination(question, retrieved_context, generated_answer):
    """
    This function acts as an AI Tester. It reads the context and the answer,
    and gives a score of 1 (Pass) or 0 (Fail) for Hallucinations.
    """
    eval_prompt = f"""
    You are an expert AI QA Tester. Your job is to measure 'Faithfulness' (Hallucination).
    Look at the CONTEXT retrieved from the database, and the ANSWER generated by the RAG bot.

    QUESTION: {question}
    CONTEXT: {retrieved_context}
    ANSWER: {generated_answer}

    Did the ANSWER invent any information that is NOT present in the CONTEXT?
    Return a JSON object with 'score' (1 for no hallucination, 0 for hallucination) and 'reason'.
    """

    try:
        response = judge_client.chat.completions.create(
            model="openai/gpt-oss-120b", # Using Llama-3 as our strict QA Judge
            messages=[{"role": "user", "content": eval_prompt}],
            response_format={"type": "json_object"},
            temperature=0 # Temp = 0 so the judge doesn't hallucinate its own scores!
        )

        # Parse the JSON score
        result = json.loads(response.choices[0].message.content)
        print("====================================")
        print("🤖 RAG EVALUATION REPORT")
        print("====================================")
        print(f"📊 Hallucination Score : {result.get('score')}/1")
        print(f"📝 Judge's Reason      : {result.get('reason')}")
        print("====================================")

        return result.get('score')

    except Exception as e:
        print(f"❌ Evaluation Error: {e}")
        return None

# --- AUTOMATED TEST EXECUTION ---

print("⏳ Running AI Evaluation...\n")

# Scenario: The bot hallucinates a fake fact
my_question = "Why is self-attention better?"
my_context = "Self-attention connects all words directly, improving processing speed."
my_answer = "Self-attention connects all words directly, and it was invented by Elon Musk in 2020."

# Run the automated test!
evaluate_rag_hallucination(my_question, my_context, my_answer)
```
